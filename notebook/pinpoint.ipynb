## 1. Clonar el proyecto


```python
URL = "https://github.com/DavidEscalanteG/PLN-Pinpoint-Project.git"
PROJECT_DIR = "PLN-Pinpoint-Project"

import os, subprocess

def clonar():
    if os.path.isdir(PROJECT_DIR):
        print(f"Ya existe una copia en ./{PROJECT_DIR}, se usa tal cual.")
        return
    result = subprocess.run(
        ["git", "clone", "--depth", "1", URL, PROJECT_DIR],
        capture_output=True, text=True,
    )
    if result.returncode == 0:
        print("Repositorio clonado correctamente.")
    else:
        print("No se pudo clonar el repositorio:")
        print(result.stderr)

clonar()
%cd {PROJECT_DIR}
```

## 3. Configurar la partida

`lang` acepta `"spa"`, `"eng"` o `"fra"`. `rounds` define cuántas categorías se juegan.
`relacion = True` muestra, junto a cada pista, de qué relación de WordNet proviene.

```python
import sys

from pinpoint.game import GameSession
from pinpoint.models import PinpointError, CategoryBankError
from pinpoint.scoring import max_points

LANG = "eng"          # Cambiar segun el idioma: "spa" | "eng" | "fra"
ROUNDS = 3
SEED = None
relacion = True

print(f"Sesión creada: {ROUNDS} ronda(s) en idioma '{LANG}'.")
```

## 4. Jugar

Esta celda no reimplementa nada: importa `cli.py` (el mismo módulo que corre `python cli.py` en terminal) y llama a su función `jugar_partida()`. Así, cualquier arreglo que se le haga a `cli.py` —como el manejo de una ronda interrumpida— se refleja aquí automáticamente, sin mantener dos copias del mismo código.

Se puede correr varias veces mientras la partida no haya terminado (`sesion.is_over`); si interrumpes a media ronda, al volver a correr esta celda retoma exactamente donde la dejaste.

```python
import cli
from pinpoint.models import Lang

sesion = GameSession(LANG, rounds=ROUNDS, seed=SEED)
lang = Lang(LANG)

print("PINPOINT WORDNET")
print(f"  Pistas en: {sesion.lang.label}  |  Responde en: {sesion.lang.label}")
print("  5 pistas por ronda, de la más general a la más específica.")
print("  Puntos: A menor número de pistas usadas mayor puntaje, pista 1 = 5 pts ... pista 5 = 1 pt.")
print("  Comandos: ':pasar' (siguiente pista), ':salir' (terminar).")


while not sesion.is_over:
    sesion.new_round()
    if not cli.jugar(sesion, relacion):
        break

cli.resultado(sesion)
```
